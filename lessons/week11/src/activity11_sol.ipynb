```python
import scipy

# TODO: import pd, np, and matplotlib libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# TODO: write down the mean of a standard normal distribution
mean = 0
# TODO: write down the std of a standard normal distribution
std = 1

# TODO: use a function out of numpy called random.normal, passing in the mean, std, and how many samples to draw
data = np.random.normal(loc=mean, scale=std, size=1000)

# TODO: convert to dataframe
df = pd.DataFrame(data, columns=["values"])

print(df.head())
df.shape
```

```python
df.mean()
```

```python
df.std()
```

```python
from scipy import stats

z_value = 1.96
percentile = stats.percentileofscore(df["values"], z_value)
print("Percentile:", percentile)
```

```python
quantile_value = 0.975
z_estimate = np.quantile(df["values"], quantile_value)
print("Estimated z-score:", z_estimate)
```

```python
alpha = 0.05

quantile_value = 1 - alpha / 2

z_estimate = np.quantile(df["values"], quantile_value)

print("Quantile (1 - alpha/2):", quantile_value)
print("Estimated z-score:", z_estimate)
```

```python
# TODO: plot the random normal data!
plt.hist(df["values"], bins=30, density=True) # makes the total area equal 1, but the y-axis shows density—not probability.
# TODO: add a vertical line on your figure to mark the critical z-value
plt.axvline(z_estimate, linestyle='--', linewidth=2)

plt.title("Standard Normal Distribution with Critical Z-Value")
plt.xlabel("Z-score")
plt.ylabel("Density")
plt.show()
```

```python
import numpy as np
from scipy import stats

# given values
mu = 500
sem = 20
middle_area = 0.80

# tail areas
alpha = 1 - middle_area
tail_area = alpha / 2

# z boundaries
z_upper = stats.norm.ppf(1 - tail_area)
z_lower = stats.norm.ppf(tail_area)

# convert z boundaries to raw sample-mean boundaries
x_lower = mu + z_lower * sem
x_upper = mu + z_upper * sem

print("alpha =", alpha)
print("tail area =", tail_area)
print("z lower =", z_lower)
print("z upper =", z_upper)
print("lower bound =", x_lower)
print("upper bound =", x_upper)
```
